Data Prep

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons,make_circles,make_classification
```

```python
X,y=make_moons(n_samples=2000,noise=0.15,random_state=1)
```

```python
W,u=make_circles(n_samples=2000,noise=0.1,random_state=1,factor=0.5)
```

```python
A,b=make_classification(n_samples=2000,n_features=2,n_redundant=0,random_state=0)
```

```python
print(X.shape,y.shape)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
plt.show()
```

```python
plt.scatter(W[:,0],W[:,1],c=u)
plt.show()
```

```python
plt.scatter(A[:,0],A[:,1],c=b)
plt.show()
```

```python
def load_dataseet(dataset):
    if dataset == 'moons':
        X,y=make_moons(n_samples=10000,noise=0.15,random_state=1)
    elif dataset == 'circles':
        X,y=make_circles(n_samples=10000,noise=0.1,random_state=1,factor=0.5)
    elif dataset == 'classification':
        X,y=make_classification(n_samples=10000,n_features=2,n_redundant=0,random_state=1)

    return X,y
```

```python
dataset='moons'
X,y=load_dataseet(dataset)
print(X.shape,y.shape)
plt.scatter(X[:,0],X[:,1],c=y)
plt.show()
```

Model Building

```python
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(10,activation='relu',input_shape=(2,)))
model.add(Dense(5,activation='relu'))
model.add(Dense(1,activation='sigmoid'))

model.compile(loss='binary_crossentropy',metrics=['accuracy'])
model.summary()
```

Model Training and Testing

```python
model.fit(X,y,epochs=50,batch_size=64,validation_split=0.1)
```
